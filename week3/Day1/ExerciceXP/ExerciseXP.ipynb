```python
#Exercise 2
import numpy as np

numbers_list=[3.14,2.17,0,1,2]

#conversion de la liste en tableau numpy

numbers_array=np.array(numbers_list)

#conversion du type float vers int

numbers_array=numbers_array.astype(int)

print(numbers_array)
```

```python
#Exercise 3 
import numpy as np

#création d'un tableau contenant les valeurs de 1 à 9

numbers_array=np.arange(1,10)

#transformation en matrice 3x3

matrix=numbers_array.reshape(3,3)

print(matrix)
```

```python
#Exercise 6 
import numpy as np

array=np.arange(0,10)

#inversion du tableau

reversed_array=array[::-1]

print(reversed_array)
```

```python
#Exercise 7 
import numpy as np

#création d'une matrice identité 4x4

identity_matrix=np.eye(4)

print(identity_matrix)
```

```python
#Exercise 8
import numpy as np

array=np.arange(0,10)

#somme des valeurs

array_sum=array.sum()

#moyenne des valeurs

array_average=array.mean()

print(f"Sum : {array_sum}")
print(f"Average : {array_average}")
```
