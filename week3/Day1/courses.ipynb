```python
#NUMPY AVANCE : 
#Le test d'hypothèse :  on fait le testd'hypothe ssi on a déjà ds données sur lesquels on a travaillé . 
#Le test d'hypothèse est une méthode statistique qui permet de prendre des décisions ou de tirer des conclusions à partir de données échantillonnées. Il s'agit d'un processus formel qui consiste à formuler une hypothèse nulle (H0) et une hypothèse alternative (H1), puis à utiliser des données pour déterminer si les résultats sont suffisamment significatifs pour rejeter l'hypothèse nulle.

#p-value: c'est la probabilité d'obtenir des résultats aussi extrêmes que ceux observés, ou plus extrêmes, si l'hypothèse nulle est vraie. Une p-value faible (généralement inférieure à un seuil de signification, comme 0.05) indique que les résultats sont peu probables sous l'hypothèse nulle, ce qui peut conduire à rejeter cette hypothèse.

#t_statistic: c'est une mesure de la différence entre les données observées et ce qui serait attendu sous l'hypothèse nulle, normalisée par la variabilité des données. Un t-statistic élevé indique une plus grande différence entre les données observées et l'hypothèse nulle, ce qui peut également conduire à rejeter cette hypothèse.

#donc dans le test on va  faire la comparaison entre la p-value et le seuil de signification pour décider si on rejette ou non l'hypothèse nulle, et on peut aussi utiliser le t-statistic pour évaluer la magnitude de la différence entre les données observées et l'hypothèse nulle.

# si la omparaison nous montre que nore p_value est inférieur au seuil de signification (Alpha) , en gros dans notrez cas , la probabiliité d'avoir 15 min est très faible , alors on rejette l'hypothèse nulle et on conclut que les données suggèrent que le temps d'engagement moyen est différent de 15 minutes. En revanche, si la p-value est supérieure au seuil de signification, on ne rejette pas l'hypothèse nulle, ce qui signifie que les données ne fournissent pas suffisamment de preuves pour conclure que le temps d'engagement moyen est différent de 15 minutes.

import numpy as np
from scipy import stats

# Sample data (engagement time in minutes)
sample_data = np.random.randint(0, 30)  # Example data

# Performing a one-sample t-test
t_statistic, p_value = stats.ttest_1samp(sample_data, 15)

# Interpreting the result
alpha = 0.05  # Significance level 
if p_value < alpha:
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```
