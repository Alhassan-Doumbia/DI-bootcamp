## 🌟 Exercise 1: Traditional vs. Modern NLP: Comparative Analysis

### 1) Table comparative

| Aspect | Traditional NLP | Modern NLP |
|---|---|---|
| Feature Engineering | Manuelle : n-grams, TF-IDF, POS tags, règles linguistiques | Automatique : les modèles apprennent les représentations directement depuis les données |
| Word Representations | Vecteurs creux (bag-of-words, TF-IDF) ou denses fixes (Word2Vec, GloVe) | Embeddings contextuels dynamiques (BERT, GPT) : le même mot a des représentations différentes selon le contexte |
| Model Architectures | HMM, CRF, SVM, arbres de décision, réseaux récurrents (LSTM, GRU) | Transformers avec self-attention, architecture encoder/decoder ou mixte |
| Training Methodology | Supervisé sur des tâches spécifiques avec peu de données annotées | Préentraînement auto-supervisé massif (MLM, CLM) puis fine-tuning sur tâche cible |
| Key Examples | NLTK, spaCy (règles), Word2Vec, fastText, LSTM-NER | BERT, GPT-2/3/4, T5, RoBERTa, XLM-R, LLaMA |
| Advantages | Interprétable, léger, peu de données nécessaires, rapide à entraîner | Performances état de l'art, gestion du contexte long, transfert multi-tâches sans ré-annotation |
| Disadvantages | Contexte limité, ingénierie coûteuse, fragile aux variations de langue | Coût computationnel élevé, boîte noire, biais héritées des données d'entraînement |

### 2) Impact du passage au NLP moderne (5-7 phrases)

Le passage au NLP moderne a radicalement transformé la scalabilité des systèmes de traitement du langage. Les représentations apprises automatiquement (embeddings contextuels) éliminent la nécessité d'une ingénierie de features fastidieuse, ce qui permet de passer d'un domaine à l'autre avec un simple fine-tuning. L'architecture Transformer est nativement parallélisable : contrairement aux LSTM qui traitent les séquences token par token, le self-attention calcule toutes les interactions simultanément, exploitant pleinement les GPU modernes. Le transfert learning permet de réutiliser un modèle préentraîné sur des milliards de tokens pour une nouvelle tâche avec seulement quelques centaines ou milliers d'exemples annotés, réduisant drastiquement le coût de collecte de données. Cette efficacité en données s'accompagne d'une meilleure généralisation : les représentations riches capturent des relations sémantiques et syntaxiques subtiles que les méthodes symboliques peinaient à modéliser. En contrepartie, la demande en puissance de calcul et en mémoire GPU a explosé, créant une barrière à l'entrée et des enjeux environnementaux (empreinte carbone de l'entraînement). Au final, le NLP moderne a démocratisé des capacités autrefois réservées aux grandes organisations disposant d'équipes d'ingénieurs NLP spécialisés.

## 🌟 Exercise 2: LLM Architecture and Application Scenarios

### BERT
- **Architecture :** Encoder-only bidirectionnel — BERT lit la séquence entière de gauche à droite ET de droite à gauche simultanément grâce au Masked Language Modeling (MLM). Il produit des représentations contextuelles riches pour chaque token, idéales pour la compréhension.
- **Application :** Classification de sentiments dans des avis clients (ex. Amazon, Yelp).
- **Why :** BERT voit tout le contexte autour d'un mot, ce qui lui permet de détecter des nuances fines (ironie, double sens) que les modèles unidirectionnels manquent.

### GPT
- **Architecture :** Decoder-only autorégressif — GPT prédit le prochain token en se basant uniquement sur les tokens précédents (causal LM). Il génère du texte de gauche à droite, token par token.
- **Application :** Assistant de code (GitHub Copilot), chatbot conversationnel, génération automatique de contenu marketing.
- **Why :** Le modèle autorégressif est parfait pour la génération ouverte : il produit des textes cohérents et fluides de longueur variable sans avoir besoin de connaître la suite.

### T5
- **Architecture :** Encoder-Decoder symétrique — T5 encapsule toutes les tâches NLP dans un cadre texte-vers-texte : la tâche est spécifiée dans le prompt ("translate English to French: ..."), l'encodeur comprend l'entrée et le décodeur génère la sortie.
- **Application :** Traduction automatique, résumé de documents longs, reformulation de questions (question generation).
- **Why :** Le cadre texte-à-texte rend T5 extrêmement versatile : un seul modèle peut gérer la traduction, le résumé et la Q&A sans changer d'architecture, juste en changeant le préfixe de la tâche.

## 🌟 Exercise 3: Benefits and Ethics of Pre-training

### Bénéfices du préentraînement

1. **Improved generalization :** Le modèle apprend des représentations universelles du langage (syntaxe, sémantique, faits du monde) qui se transfèrent à des tâches jamais vues durant le préentraînement. Il généralise mieux qu'un modèle entraîné uniquement sur des données annotées limitées.

2. **Less labeled data :** Le préentraînement étant auto-supervisé (pas besoin d'étiquettes humaines), un fine-tuning efficace peut se faire avec quelques centaines ou milliers d'exemples annotés au lieu de millions, réduisant drastiquement le coût de création de datasets.

3. **Faster fine-tuning :** Les poids préentraînés constituent un point de départ excellent : la convergence est rapide (quelques epochs suffisent) car le modèle connaît déjà la structure du langage, il n'a qu'à s'adapter à la tâche cible.

4. **Transfer learning :** Les connaissances encodées dans un modèle préentraîné (relations entre mots, structure grammaticale, entités nommées) se transfèrent directement. Un modèle préentraîné sur de l'anglais général peut être fine-tuné pour de l'anglais médical avec peu de données médicales.

5. **Robustness :** Exposé à des données très diversifiées (news, Wikipedia, livres, forums), le modèle préentraîné est plus robuste aux variations orthographiques, aux argots et aux formulations inhabituelles que des modèles entraînés sur un corpus étroit.

### Préoccupations éthiques
- **Biais :** Les LLMs héritent des biais présents dans les données d'entraînement (stéréotypes de genre, racisme, représentations culturelles déséquilibrées). Exemple : associer certains métiers à des genres spécifiques.
- **Désinformation :** Les modèles génératifs peuvent produire des textes faux mais convaincants (hallucinations factuelles, fake news, deepfake text).
- **Mauvais usage :** Génération de contenu malveillant (spam, hameçonnage, propagande automatisée, code malveillant).
- **Vie privée :** Les données d'entraînement peuvent contenir des informations personnelles qui ressortent lors de la génération.

### Atténuations
- **Curation des données :** Filtrer activement les contenus toxiques, biaisés ou confidentiels avant le préentraînement (ex. C4, The Pile avec filtres).
- **Confidentialité différentielle :** Entraîner avec du differential privacy pour éviter la mémorisation de données personnelles.
- **Filtres de sécurité :** Systèmes de détection de contenu nuisible en sortie (modération API).
- **RLHF (Reinforcement Learning from Human Feedback) :** Aligner le modèle sur des valeurs humaines via des annotations de préférence.
- **Audits réguliers :** Évaluation externe des biais et capacités nuisibles sur des benchmarks dédiés (BBQ, TruthfulQA).

## 🌟 Exercise 4: Transformer Architecture Deep Dive

### Self-Attention & Multi-Head Attention

**Flux Q/K/V :** Chaque token d'entrée est projeté en trois vecteurs : Query (ce que le token cherche), Key (ce qu'il offre aux autres) et Value (son contenu réel). On calcule les scores d'attention par `QKᵀ / √d_k`, on applique softmax pour obtenir des poids, puis on pondère les Values. Le résultat est une représentation enrichie de chaque token par son contexte.

**Pourquoi plusieurs têtes :** Avec une seule tête d'attention, le modèle ne peut capturer qu'un seul type de relation à la fois. Plusieurs têtes projettent les Q/K/V dans des sous-espaces différents, permettant à chaque tête de spécialiser sur des relations distinctes (syntaxe, coréférence, sémantique). Leurs sorties sont concaténées et reprojetées.

**Exemple de phrase :** *"The bank on the river bank approved the loan."*
- **Tête 1** pourrait capturer la **relation sujet-verbe** : relier "bank (institution)" → "approved"
- **Tête 2** pourrait capturer la **désambiguïsation lexicale** : distinguer "bank (institution)" de "bank (rive)" en regardant le contexte "river" vs "approved"

---

### Pre-training Objectives

- **MLM (Masked Language Modeling) vs CLM (Causal Language Modeling) :** MLM masque ~15% des tokens et les prédit en utilisant tout le contexte (bidirectionnel) — idéal pour la compréhension (BERT). CLM prédit le token suivant avec uniquement le contexte gauche (unidirectionnel) — idéal pour la génération (GPT).
- **Quand préférer MLM :** Tâches de compréhension (classification, NER, Q&A extractive) où on a besoin du contexte global.
- **Quand préférer CLM :** Tâches de génération (complétion de texte, dialogue, résumé) où le modèle doit produire du texte séquentiellement.
- **NSP (Next Sentence Prediction) :** BERT original l'utilisait pour apprendre les relations inter-phrases. Mais les études ont montré que NSP apporte peu de gain et que son signal est trop facile à apprendre ; RoBERTa l'a supprimé et obtenu de meilleures performances, montrant que l'entraînement MLM seul avec plus de données est plus efficace.

---

### Transformer Model Selection

- **Analyse de sentiment sur des avis :** **Encoder-only (BERT/RoBERTa)** — la classification nécessite de comprendre l'ensemble du texte avant de décider ; la bidirectionnalité est cruciale pour détecter la négation et l'ironie.
- **Chatbot conversationnel (réponses créatives) :** **Decoder-only (GPT)** — la génération libre, fluide et contextuelle est le point fort des modèles autorégressifs ; ils maintiennent la cohérence sur de longues conversations.
- **Traduction de documents techniques (EN→ES) :** **Encoder-Decoder (T5, mBART, NLLB)** — la traduction est une transformation séquence-vers-séquence ; l'encodeur comprend la structure source et le décodeur génère la cible dans la langue cible avec alignement précis.

---

### Positional Encoding

**Pourquoi :** Le self-attention est invariant à la position par construction (QKᵀ ne dépend pas de l'ordre). Sans encoding positionnel, "Le chat mange la souris" et "La souris mange le chat" auraient les mêmes représentations. L'encoding positionnel injecte l'information d'ordre (absolue ou relative) dans les embeddings.

**Exemple d'échec sans positions :** La phrase "She gave him a book before she left" — sans positions, le modèle ne saurait pas que "she" dans "she left" réfère à la même personne que le premier "she", car l'ordre des pronoms serait perdu.

## 🌟 Exercise 5: BERT Variations: Choose Your Detective

### Affectation par scénario
- **Scénario 1 (mobile, ressources limitées) :** **DistilBERT** — 40% plus léger que BERT, 60% plus rapide, conserve 97% des performances. Parfait pour les appareils à faibles ressources.
- **Scénario 2 (documents légaux, haute précision) :** **RoBERTa** — entraîné plus longtemps, avec plus de données, sans NSP et avec des batches dynamiques. Meilleures performances absolues sur les tâches de compréhension complexe.
- **Scénario 3 (support multilingue) :** **XLM-R** — préentraîné sur 100 langues avec une architecture RoBERTa robuste. Excellentes performances cross-linguales, y compris sur des langues à faibles ressources.
- **Scénario 4 (détection remplacement de tokens, entraînement efficace) :** **ELECTRA** — discriminateur qui détecte si un token a été remplacé (vs MLM qui masque). 4× plus efficace en calcul pour le même niveau de performance.
- **Scénario 5 (NLP efficace, environnements contraints) :** **ALBERT** — partage les paramètres entre les couches et décompose la matrice d'embedding. Beaucoup moins de paramètres tout en maintenant des performances comparables à BERT.

## 🌟 Exercise 6: Softmax Temperature: The Randomness Regulator

### 1) Scénarios de température

- **T = 0.2 :** La distribution softmax est très pointue (peaky) : presque toute la probabilité se concentre sur le token le plus probable. Le modèle génère du texte très **déterministe et répétitif**. Utile pour des tâches où la précision prime (extraction d'informations, résumé factuel).

- **T = 1.5 :** La distribution est très **aplatie** : les tokens peu probables ont une chance non négligeable d'être choisis. Le texte devient plus **créatif mais incohérent** — le modèle peut sortir des sentiers battus mais produit aussi plus d'erreurs et de non-sens.

- **T = 1.0 :** La distribution est **identique à celle apprise** pendant l'entraînement. C'est le comportement nominal : bonne balance entre cohérence et diversité. Les tokens sont échantillonnés selon les probabilités réelles du modèle.

### 2) Application Design

- **Histoires pour enfants (créativité vs cohérence) :** Utiliser une **température modérément haute** (~0.9–1.1) combinée avec `top_p=0.9`. On veut de la créativité narrative (rebondissements inattendus, descriptions imagées) mais les phrases doivent rester cohérentes et lisibles pour les enfants. Une température trop haute produirait des absurdités.

- **Résumés de rapports financiers (précision/fiabilité) :** Utiliser une **température très basse** (~0.1–0.3) avec **greedy decoding ou beam search**. Les chiffres, pourcentages et faits doivent être reproduits fidèlement — toute hallucination créative serait dangereuse dans un contexte financier. On préfère la répétabilité à la diversité.

### 3) Température & Biais

La température influence l'amplification ou l'atténuation des biais. **À basse température**, le modèle suit étroitement les patterns les plus fréquents dans ses données d'entraînement — or ces patterns reflètent les biais sociaux dominants. Par exemple, à T=0.2, compléter "The doctor said that ___ was tired" donnera systématiquement "he" si le corpus sur-représente les médecins masculins.

**À haute température**, la distribution est plus diffuse et le modèle échantillonne parfois des tokens moins fréquents — ce qui peut en apparence "diluer" les biais mais en réalité les rend imprévisibles et potentiellement plus choquants (termes offensants minoritaires remontant aléatoirement). La solution n'est pas dans la température mais dans la curation des données et l'alignement par RLHF.
