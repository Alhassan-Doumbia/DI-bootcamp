```python
#Exercise 1 
import requests
import pandas as pd 
#le dataset a été téléchargé sur kaggle , voilà le lien : https://www.kaggle.com/datasets/arunjangir245/boston-housing-dataset/code

def Analysis_of_column(column_name="",datasetName=""):
    df=pd.read_csv(f'{datasetName}')
    # #debugging : voir si le dataset est bien chargé
    # print(df)
    columns=list(df.columns)
    if column_name in columns:
        try: 
            print(f"The mean value of the column {column_name} is : {df[f'{column_name}'].mean()}")
            # return the median value
            print(f"The median value of the column {column_name} is : {df[f'{column_name}'].median()}")
            # return standard deviation "ecart type"
            print(f"The standard deviation of the column {column_name} is : {df[f'{column_name}'].std()}")
        except Exception as e : 
            print(f"there is a problem, check : {e}")
            print("end of task")
    else:
        print("The column is not in the dataset")


Analysis_of_column("age","BostonHousing.csv") 

```
