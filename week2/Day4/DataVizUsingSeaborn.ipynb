```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np


#loafing sample data 
data=sns.load_dataset('tips')
sns.histplot(data['total_bill'], bins=20, color='skyblue')
plt.title('Histogram of Total Bill')
plt.show()


```

```python
sns.kdeplot(data['total_bill'], shade=True, color='purple')
plt.title('KDE of Total Bill')
plt.show()

#Graphiques d'estimation de la densité du noyau (KDE)
```

```python
#jointplot :permet de joindre deux graphique afin de visualiser la relation entre deux variables
sns.jointplot(x='total_bill',y='tip',data=data,kind='scatter')
plt.show()
```
