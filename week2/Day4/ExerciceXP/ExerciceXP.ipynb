## Exercise 1 : Understanding Data Visualization

### Pourquoi la visualisation de données est-elle importante ?

La visualisation de données permet de **comprendre rapidement** de grandes quantités d'informations. Un graphique bien construit révèle en quelques secondes des tendances, des anomalies ou des relations qui prendraient des heures à identifier dans un tableau de chiffres bruts.

Elle sert à plusieurs choses en data analysis :
- **Explorer** les données (trouver des patterns, des valeurs aberrantes)
- **Communiquer** des résultats à des personnes non techniques
- **Valider** des hypothèses statistiques de façon intuitive

### Quel est le rôle d'un graphique en ligne (line graph) ?

Un **graphique en ligne** est utilisé pour représenter l'**évolution d'une valeur dans le temps** (ou selon une variable continue). Chaque point représente une valeur, et la ligne qui relie les points permet de voir la tendance globale : hausse, baisse, stabilité, cycles saisonniers, etc.

Exemple typique : l'évolution de la température sur une semaine, ou les ventes mensuelles d'une entreprise.

## Exercise 2 : Line Plot – Température sur une semaine

```python
import matplotlib.pyplot as plt

days         = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
temperatures = [72, 74, 76, 80, 82, 78, 75]

plt.figure(figsize=(9, 5))
plt.plot(days, temperatures, marker='o', color='steelblue', linewidth=2, markersize=7)

plt.xlabel('Day')
plt.ylabel('Temperature (°F)')
plt.title('Temperature Variation Over a Week')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Exercise 3 : Bar Chart – Ventes mensuelles

```python
import matplotlib.pyplot as plt

months = ['January', 'February', 'March', 'April', 'May']
sales  = [5000, 5500, 6200, 7000, 7500]

plt.figure(figsize=(9, 5))
plt.bar(months, sales, color='coral', edgecolor='white')

plt.xlabel('Month')
plt.ylabel('Sales Amount ($)')
plt.title('Monthly Sales for a Retail Store')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

## Exercise 4 : Histogramme – Distribution des CGPA des étudiants

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('Student Mental health.csv')

print('Colonnes :', df.columns.tolist())
print('Valeurs CGPA :', df['What is your CGPA?'].value_counts())

plt.figure(figsize=(10, 5))
sns.histplot(data=df, x='What is your CGPA?', color='steelblue', shrink=0.8)

plt.title('Distribution of Student CGPA')
plt.xlabel('CGPA Range')
plt.ylabel('Number of Students')
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

## Exercise 5 : Bar Plot – Niveaux d'anxiété par genre

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# df déjà chargé à l'exercice précédent, on recharge pour être sûr
df = pd.read_csv('Student Mental health.csv')

# Vérifier les noms exacts des colonnes
print(df[['Do you have Anxiety?', 'Choose your gender']].value_counts())

plt.figure(figsize=(8, 5))
sns.countplot(
    data=df,
    x='Choose your gender',
    hue='Do you have Anxiety?',
    palette='Set2'
)

plt.title('Anxiety Levels Across Different Genders')
plt.xlabel('Gender')
plt.ylabel('Number of Students')
plt.legend(title='Anxiety')
plt.tight_layout()
plt.show()
```

## Exercise 6 : Scatter Plot – Âge et crises de panique

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('Student Mental health.csv')

# Convertir la réponse Yes/No en 1/0
df['panic_numeric'] = df['Do you have Panic attack?'].map({'Yes': 1, 'No': 0})

plt.figure(figsize=(10, 5))
sns.scatterplot(
    data=df,
    x='Age',
    y='panic_numeric',
    hue='Do you have Panic attack?',
    palette={'Yes': 'crimson', 'No': 'steelblue'},
    s=80,
    alpha=0.7
)

plt.title('Relationship Between Age and Panic Attacks')
plt.xlabel('Age')
plt.ylabel('Panic Attack (1 = Yes, 0 = No)')
plt.yticks([0, 1], ['No', 'Yes'])
plt.legend(title='Panic Attack')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# Résumé rapide
print('Nombre de crises de panique par âge :')
print(df.groupby('Age')['panic_numeric'].sum().sort_values(ascending=False).head(10))
```
