```python
#Apprendre  à utiliser matplotlib 
import matplotlib.pyplot as plt
import numpy as np


X=np.linspace(-5,5,100)
Y=np.power(X,2)
plt.plot(X,Y)
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Plot of Y = X^2')

plt.show()
```

```python
brands=['Brand A', 'Brand B', 'Brand C', 'Brand D']
sales=np.random.randint(100,1000,size=len(brands))

plt.bar(brands,sales,color=['blue','orange','green','red'])
plt.xlabel('brands')
plt.ylabel('sales')
plt.title('sales of different brands')

plt.show()
```

```python
data=np.random.normal(16,10,1000)
plt.hist(data,bins=10,color='purple',edgecolor='black') #bins represente le nombre de barres (et les bas) que je veux dans mon histogramme , edgecolor c'est la couleur des bordures des barres
plt.xlabel('value')
plt.ylabel('frequency')
plt.title('Histogram of Normally Distributed Data')
plt.show()
```
