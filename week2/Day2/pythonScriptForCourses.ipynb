```python
import pandas as pd

# 1. Création du DataFrame des clients
data_clients = {
    'client_id': [101, 102, 103, 104],
    'nom': ['Sophie', 'Lucas', 'Emma', 'Arthur']
}
df_clients = pd.DataFrame(data_clients)

# 2. Création du DataFrame des commandes
data_commandes = {
    'commande_id': [1, 2, 3, 4],
    'client_id': [101, 103, 101, 102],
    'montant': [50.50, 120.00, 35.00, 89.90]
}
df_commandes = pd.DataFrame(data_commandes)

print("--- Table Clients ---")
print(df_clients)
print("\n--- Table Commandes ---")
print(df_commandes)


# Fusion des deux DataFrames sur la colonne commune 'client_id'
df_fusion = pd.merge(df_clients, df_commandes, on='client_id')

print("\n--- Résultat de la fusion (Inner Join) ---")
print(df_fusion)


df_left = pd.merge(df_clients, df_commandes, on='client_id', how='left')
print("\n--- Résultat de la fusion (Left Join) ---")
print(df_left)
```

```python
import pandas as pd
#load the Titanic the dataset 
titanic_data=pd.read_csv('train.csv')

#explore the dataset 
# print(titanic_data.head(5))
print(titanic_data.info())

# Remove irrelevant columns (e.g., 'Cabin' and 'Ticket')
titanic_data = titanic_data.drop(['Cabin', 'Ticket'], axis=1)

print(titanic_data.head(5))

```

```python
Q1=titanic_data['Age'].quantile(0.25)
Q3=titanic_data['Age'].quantile(0.75)
IQR=Q3-Q1  

lower_bound=Q1-1.5*IQR
upper_bound=Q3+1.5*IQR

print(f"Q1: {Q1}, Q3: {Q3}, IQR: {IQR}")
print(f"Lower Bound: {lower_bound}, Upper Bound: {upper_bound}")

titanic_data=titanic_data[(titanic_data['Age']>=lower_bound) & (titanic_data['Age']<=upper_bound)]
print(titanic_data.info())
```

```python
#Data normalization 

titanic_data.head(5)
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
titanic_data['Age_normalized'] = scaler.fit_transform(titanic_data[['Age']])

print(titanic_data[['Age', 'Age_normalized']].head(5))
```
