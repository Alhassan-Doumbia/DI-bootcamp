```python

#Exercise 3
from scipy import stats
import numpy as np

data = [12, 15, 13, 12, 18, 20, 22, 21]

mean = np.mean(data)
median = np.median(data)
variance = np.var(data)
std = np.std(data)

print("Moyenne :", mean)
print("Médiane :", median)
print("Variance :", variance)
print("Écart-type :", std)
```

```python
#Exercise 3

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

x = np.linspace(20, 80, 100)

y = norm.pdf(x, loc=50, scale=10)

plt.plot(x, y)
plt.title("Distribution Normale")
plt.xlabel("Valeurs")
plt.ylabel("Densité")
plt.show()
```

```python
#eXERCISE 4

import numpy as np
from scipy.stats import ttest_ind

data1 = np.random.normal(50, 10, 100)
data2 = np.random.normal(60, 10, 100)

t_stat, p_value = ttest_ind(data1, data2)

print("T-statistic :", t_stat)
print("P-value :", p_value)

if p_value < 0.05:
    print("Différence significative")
else:
    print("Pas de différence significative")
```

```python
#Exercice 5
from scipy.stats import linregress

house_sizes = [50, 70, 80, 100, 120]
house_prices = [150000, 200000, 210000, 250000, 280000]

result = linregress(house_sizes, house_prices)

print("Slope :", result.slope)
print("Intercept :", result.intercept)

prix_90 = result.slope * 90 + result.intercept

print("Prix prédit pour 90 m² :", prix_90)
```

```python
#Exercice 6
from scipy.stats import f_oneway

fertilizer_1 = [5, 6, 7, 6, 5]
fertilizer_2 = [7, 8, 7, 9, 8]
fertilizer_3 = [4, 5, 4, 3, 4]

f_stat, p_value = f_oneway(
    fertilizer_1,
    fertilizer_2,
    fertilizer_3
)

print("F-value :", f_stat)
print("P-value :", p_value)

if p_value < 0.05:
    print("Les engrais ont des effets différents")
else:
    print("Pas de différence significative")
```

```python
#Exercice 7
from scipy.stats import binom

n = 10
p = 0.5

prob = binom.pmf(5, n, p)

print("Probabilité :", prob)
```

```python
#Exercise 8
import pandas as pd
from scipy.stats import pearsonr, spearmanr

data = pd.DataFrame({
    'age': [23, 25, 30, 35, 40],
    'income': [35000, 40000, 50000, 60000, 70000]
})

pearson_corr, p1 = pearsonr(data["age"], data["income"])
spearman_corr, p2 = spearmanr(data["age"], data["income"])

print("Pearson :", pearson_corr)
print("Spearman :", spearman_corr)
```
