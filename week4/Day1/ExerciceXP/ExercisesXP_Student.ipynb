# Loan Default Prediction: Problem Statement and Data Collection Plan

## Problem Statement

The objective of this project is to predict whether a borrower will default on a loan or not. The target variable is **loan default status** (default or no default). This prediction can help financial institutions make better lending decisions and reduce the risk of financial losses.

## Data Types Needed

### Personal Information

I would collect basic information about applicants such as age, gender, marital status, education level, and employment status. This information can help identify patterns between personal characteristics and the likelihood of default.

### Financial Information

I would collect data about income, monthly expenses, savings, and existing debts. These variables are important because they provide information about the applicant’s financial capacity to repay the loan.

### Credit History

I would collect credit scores, previous loans, and records of late payments. Credit history is often one of the strongest indicators of whether a person is likely to repay a loan on time.

### Loan Information

I would collect information about the loan amount, interest rate, loan duration, and loan purpose. These factors can influence the difficulty of repayment and therefore the probability of default.

### Repayment History

For customers who have already borrowed from the institution, I would collect repayment records and payment delays. Past repayment behavior can help predict future repayment behavior.

## Data Sources

### Financial Institution Records

Most of the information can be obtained from the bank or financial institution's internal database. This data is usually collected during the loan application process and stored in customer records.

### Credit Bureaus

Credit bureaus can provide credit scores and information about previous loans. The financial institution can integrate this data through data-sharing agreements and authorized credit checks.

### Customer Applications

Applicants can provide personal and financial information directly through loan application forms. This data can be collected digitally or through paper forms and then stored in a central database.

### Transaction Records

Bank account transactions and payment histories can also be useful sources of information. These records can be integrated from the institution's existing banking systems.

## Risks and Constraints

Several risks and constraints must be considered during this project. Personal financial data is sensitive and must be protected according to privacy and data protection regulations. Data quality can also be a challenge because some records may be incomplete, outdated, or incorrect. Another risk is sampling bias if the dataset does not represent all types of borrowers. Finally, proper data governance is important to ensure that data is used ethically, securely, and only for authorized purposes.

## Feature Selection Justification

### Selected Features

I selected the following features as the most relevant for predicting loan defaults: **credit_score**, **debt_to_income**, **annual_income**, **loan_amount**, **interest_rate**, **employment_length**, **num_delinquencies**, **total_utilization**, **home_ownership**, **purpose**, and **term**.

**credit_score** is one of the strongest predictors because it summarizes the borrower's entire credit history into a single number that lenders have used for decades to assess repayment risk.

**debt_to_income** measures how much of a person's monthly income is already going to debt payments; a high ratio means the borrower has little financial room left, making default more likely.

**annual_income** indicates the borrower's capacity to repay: a higher income generally means the borrower can handle loan payments more comfortably.

**loan_amount** matters because larger loans require larger monthly payments, which increases the financial burden on the borrower and therefore the risk of default.

**interest_rate** directly affects the monthly payment size; higher interest rates usually signal that the lender already considered the borrower risky, and they also increase the total repayment burden.

**employment_length** reflects job stability: borrowers who have been employed for a longer period are generally considered more financially stable.

**num_delinquencies** captures the number of past payment failures, which is a direct behavioral signal of repayment habits.

**total_utilization** shows how much of the available credit the borrower is already using; a high utilization rate suggests financial stress.

**home_ownership** provides context about financial stability, since homeowners tend to have more financial assets and a stronger motivation to maintain good credit.

**purpose** helps identify higher-risk loan categories, as some loan purposes (for example, debt consolidation) tend to be associated with higher default rates than others.

**term** is important because longer loan terms carry more uncertainty and are associated with higher cumulative interest, which can increase repayment difficulty.

### Excluded Features

I excluded **age** because it can introduce demographic bias and in many jurisdictions it is legally prohibited to use age as a lending criterion.

I excluded **num_open_accounts** because this information is already partially captured by **total_utilization** and **debt_to_income**, making it redundant.

I excluded **state** and **zip_code** because these geographic identifiers can act as proxies for protected characteristics and may introduce unfair bias without adding enough predictive value to justify their inclusion.

I excluded **application_type** because in this dataset it does not provide enough direct information about the borrower's financial behavior to be useful.

### Encoding and Imputation

For categorical features such as **home_ownership**, **purpose**, and **term**, I would use one-hot encoding so that the model treats each category independently without implying any numerical order between them.

For missing values in numerical features such as **annual_income** and **employment_length**, I would impute using the median rather than the mean, because the median is more robust to outliers that are common in financial data.

```python

# This piece of code is already prefilled, run it to execute it and see the results.
# It demonstrates standard classification metrics for binary loan default prediction.

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, average_precision_score, confusion_matrix, classification_report

# Please replace these placeholders with your true labels and predicted probabilities.
y_true = [0,1,0,1,0,0,1,0,1,0]            # placeholder labels
y_pred_proba = [0.05,0.80,0.10,0.65,0.20,0.15,0.70,0.30,0.85,0.25]  # placeholder probabilities

# You should set a decision threshold that reflects the precision–recall trade-off for your business case.
threshold = 0.5
y_pred = [1 if p >= threshold else 0 for p in y_pred_proba]

print("Accuracy:", round(accuracy_score(y_true, y_pred), 4))
print("Precision:", round(precision_score(y_true, y_pred, zero_division=0), 4))
print("Recall:", round(recall_score(y_true, y_pred, zero_division=0), 4))
print("F1-score:", round(f1_score(y_true, y_pred, zero_division=0), 4))
print("ROC-AUC:", round(roc_auc_score(y_true, y_pred_proba), 4))
print("PR-AUC (Average Precision):", round(average_precision_score(y_true, y_pred_proba), 4))
print("\nConfusion matrix:\n", confusion_matrix(y_true, y_pred))
print("\nClassification report:\n", classification_report(y_true, y_pred, zero_division=0))
```

## Model Choice, Evaluation, Class Imbalance, and Threshold Strategy

### Model Choice

I would start with **Logistic Regression** as a baseline model because it is easy to interpret, runs quickly, and gives probability outputs that are well-suited for a binary classification problem like loan default prediction. I would also train a **Random Forest** classifier as a stronger candidate because it handles non-linear relationships between features, is robust to outliers, and naturally provides feature importance scores that help explain the predictions to business stakeholders.

### Evaluation Plan

I would split the data into 80% for training and 20% for testing, making sure to use stratified splitting so that the proportion of defaults is preserved in both sets. I would then apply **5-fold stratified cross-validation** on the training set to get a reliable estimate of model performance before touching the test set. The main metrics I would report are **ROC-AUC** to measure the model's overall ability to separate defaulters from non-defaulters, **PR-AUC (Average Precision)** because it is more informative than ROC-AUC when the classes are imbalanced, **Recall (Sensitivity)** because missing a true defaulter is costly for the bank, **Precision** to avoid flagging too many good borrowers as risky, and **F1-score** as a balanced measure of both. I would also examine the **confusion matrix** to understand the actual counts of false positives and false negatives.

### Addressing Class Imbalance

Loan default datasets are typically imbalanced because most borrowers repay their loans. To handle this, I would use **stratified splits** during cross-validation so each fold maintains the original class ratio. In addition, I would set `class_weight='balanced'` in both Logistic Regression and Random Forest, which automatically gives more weight to the minority class (defaulters) during training without needing to resample the data.

### Threshold Selection

By default, scikit-learn classifiers use a threshold of 0.5, but for loan default prediction it is often better to lower this threshold to increase recall, since the business cost of missing a defaulter is higher than the cost of incorrectly flagging a non-defaulter. I would plot the Precision-Recall curve and choose a threshold that achieves a good balance between recall and precision based on the acceptable risk level agreed upon with the business team.

## Machine Learning Paradigms for Specific Scenarios

### Scenario 1 : Predicting Stock Prices

The appropriate machine learning paradigm for predicting stock prices is **supervised learning**, specifically a **regression** task. The input data would be historical features such as past closing prices, trading volume, moving averages, and macroeconomic indicators. The output is a continuous numerical value representing the predicted future price of the stock. The learning signal comes from minimizing the error between the model's predicted prices and the actual historical prices observed in the training data, for example using Mean Squared Error as the loss function.

### Scenario 2 : Organizing a Library of Books

The appropriate paradigm for grouping books into genres or categories is **unsupervised learning**, specifically a **clustering** algorithm such as K-Means or hierarchical clustering. The input data would be numerical representations of the book content, such as TF-IDF vectors of the book descriptions or titles, or word embeddings. The output is a set of clusters where each book is assigned to a group with similar books, without any predefined labels. The learning objective is to minimize the within-cluster distance so that books in the same group are as similar as possible to each other.

### Scenario 3 : Robot Navigation in a Maze

The output is an action the robot should take at each step, such as moving left, right, up, or down. The learning signal is a reward function: the robot receives a positive reward when it reaches the exit and a small negative reward for each step taken, which encourages it to find the shortest path as quickly as possible.

## Evaluation Strategy for Three Types of ML Models

### Supervised Learning : Classification Model (Logistic Regression)

For a supervised classification model used to predict loan defaults, I would use **5-fold stratified cross-validation** to evaluate performance reliably, ensuring that each fold contains a representative proportion of the minority class. I would report **Accuracy** as a general overview, but I would focus primarily on **Precision**, **Recall**, **F1-score**, and **ROC-AUC** because they give a more complete picture of performance when classes are imbalanced. I would also plot the **ROC curve** to visualize the trade-off between the true positive rate and false positive rate at different thresholds, and the **Precision-Recall curve** to understand how recall and precision change together. A key challenge in evaluating supervised models is **data leakage**: if any information from the test set accidentally influences training or preprocessing, the evaluation metrics will be overly optimistic and will not reflect real-world performance.

### Unsupervised Learning : Clustering Model (K-Means)

For a clustering model used to group books by genre, I would assess cluster quality using the **Silhouette Score**, which measures how similar each data point is to its own cluster compared to other clusters; a score close to 1 indicates well-separated clusters. I would also apply the **Elbow Method** by plotting the within-cluster sum of squares (inertia) against the number of clusters K and looking for the point where the inertia curve bends sharply, which suggests the optimal number of clusters. Additionally, I would visualize the clusters using dimensionality reduction techniques such as **PCA** or **t-SNE** to check whether the clusters are visually distinct and interpretable. A key challenge in evaluating unsupervised models is that **there is no ground truth label**, which makes it impossible to know objectively whether the clusters are correct; evaluation depends on internal metrics and domain knowledge.

### Reinforcement Learning : Maze Navigation Agent

For a reinforcement learning agent trained to navigate a maze, I would measure success primarily through the **cumulative reward** collected over each episode, plotting it over training time to see whether it increases and eventually stabilizes. I would assess **convergence** by checking whether the cumulative reward and the length of each episode (number of steps to reach the exit) stop improving significantly after a certain number of training episodes, which suggests the agent has learned a near-optimal policy. I would also monitor the **exploration vs. exploitation balance** by tracking the epsilon value if using an epsilon-greedy strategy: early in training the agent should explore widely, and later it should exploit its learned knowledge. A key challenge in evaluating reinforcement learning models is that **convergence can be unstable**: the agent's performance can suddenly drop even after it has previously learned a good policy, especially if the learning rate or exploration schedule is not carefully tuned.
