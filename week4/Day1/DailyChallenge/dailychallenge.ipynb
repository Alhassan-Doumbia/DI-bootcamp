## 1. Qu'est-ce que le Machine Learning ?

Le **Machine Learning (apprentissage automatique)** est une branche de l'intelligence artificielle qui permet à un système d'apprendre des **patterns** à partir de données, sans être explicitement programmé pour chaque règle. Au lieu d'écrire des instructions fixes (`si X alors Y`), on entraîne un algorithme sur des exemples pour qu'il découvre lui-même la relation entre les données d'entrée et le résultat attendu, puis qu'il généralise cette relation à de nouvelles données jamais vues.

### Pourquoi est-ce important pour un(e) data analyst ?

- **Aller au-delà de la description** : l'analyse classique (dashboards, KPIs, statistiques descriptives) explique *ce qui s'est passé*. Le ML permet de **prédire** ce qui va se passer (churn, ventes, pannes) et de **détecter des patterns invisibles** à l'œil nu dans de gros volumes de données.
- **Automatiser des décisions répétitives** : classification de tickets, détection de fraude, scoring de leads — des tâches trop volumineuses pour être traitées manuellement.
- **Exploiter la donnée non structurée** : texte, images, logs — le ML permet d'en extraire de la valeur, ce que les méthodes statistiques classiques gèrent difficilement.
- **Améliorer la prise de décision** : en fournissant des scores de probabilité et des recommandations chiffrées plutôt que des intuitions.

### Applications du Machine Learning par industrie

| Industrie | Exemple d'application |
|---|---|
| **Santé** | Prédiction de maladies (ex. diabète, cancer) à partir de données médicales, aide au diagnostic par imagerie |

| **E-commerce / Retail** | Systèmes de recommandation de produits (ex. Amazon, Netflix), prévision de la demande et gestion des stocks |

## 2. Les trois grands types de Machine Learning

### 🔵 Supervised Learning (Apprentissage supervisé)

**Définition :** le modèle apprend à partir de données **étiquetées** (labellisées) : chaque exemple d'entraînement est composé d'une entrée `X` et de la sortie attendue `y`. Le but est d'apprendre la fonction `f` telle que `f(X) ≈ y`, afin de prédire `y` pour de nouvelles entrées.

- Deux sous-catégories : **classification** (sortie catégorielle, ex. spam/non-spam) et **régression** (sortie continue, ex. prix d'une maison).

**Exemple d'application :** prédire si un e-mail est un spam ou non, à partir d'un historique d'e-mails déjà étiquetés "spam" / "non spam".

### 🟢 Unsupervised Learning (Apprentissage non supervisé)

**Définition :** le modèle travaille sur des données **non étiquetées** : il n'y a pas de sortie `y` à prédire. L'objectif est de découvrir une **structure cachée** dans les données (groupes, associations, réduction de dimension).

- Exemples de techniques : clustering (ex. K-Means), réduction de dimensionnalité (ex. PCA), détection d'anomalies.

**Exemple d'application :** segmenter une base de clients en groupes homogènes (ex. "gros acheteurs occasionnels" vs "petits acheteurs réguliers") sans connaître les segments à l'avance, afin d'adapter les campagnes marketing.

### 🟠 Reinforcement Learning (Apprentissage par renforcement)

**Définition :** un **agent** apprend à prendre des décisions en interagissant avec un **environnement**. Il reçoit une **récompense** (positive ou négative) selon la qualité de ses actions, et ajuste sa stratégie (policy) pour maximiser la récompense cumulée sur le long terme. Il n'y a ni données étiquetées, ni structure à découvrir : l'apprentissage se fait par essais-erreurs.

**Exemple d'application :** un agent qui apprend à jouer aux échecs ou à un jeu vidéo en jouant des milliers de parties, ou un robot qui apprend à se déplacer en évitant les obstacles.

### Tableau comparatif

| | Données | Objectif | Exemple |
|---|---|---|---|
| **Supervisé** | Étiquetées (X, y) | Prédire une sortie connue | Détection de spam |
| **Non supervisé** | Non étiquetées (X seul) | Découvrir une structure cachée | Segmentation client |
| **Renforcement** | Interactions + récompenses | Maximiser une récompense cumulée | Agent qui apprend à jouer à un jeu |

## 3. Le processus de développement d'un modèle de Machine Learning

Le développement d'un modèle suit généralement trois grandes étapes (en plus de la collecte et du nettoyage des données en amont) :

### 🔹 Étape 1 : Feature Selection (Sélection des features)

C'est le choix des **variables explicatives** (features) qui seront utilisées pour entraîner le modèle.

- On identifie les colonnes pertinentes pour prédire la cible, et on écarte celles qui sont redondantes, non informatives, ou trop corrélées entre elles.
- On peut aussi **créer de nouvelles features** (feature engineering) à partir des données brutes (ex. extraire le mois à partir d'une date).
- Une bonne sélection de features réduit le bruit, accélère l'entraînement et limite le risque de surapprentissage (overfitting).

### 🔹 Étape 2 : Model Selection (Sélection du modèle)

C'est le choix de l'**algorithme** le plus adapté au problème (classification, régression, clustering...) et aux données disponibles.

- On tient compte de la taille du dataset, de la complexité des relations à modéliser, du besoin d'interprétabilité, et du temps de calcul disponible.
- Exemples : régression logistique (simple, interprétable) vs Random Forest ou réseaux de neurones (plus complexes, souvent plus performants sur des données volumineuses).
- On sépare les données en un jeu d'**entraînement** (train) et un jeu de **test** (souvent aussi un jeu de validation) pour comparer plusieurs modèles de manière équitable.

### 🔹 Étape 3 : Model Evaluation (Évaluation du modèle)

C'est la mesure de la **performance** du modèle entraîné sur des données qu'il n'a jamais vues (le jeu de test).

- Pour la classification : accuracy, precision, recall, F1-score, AUC-ROC.
- Pour la régression : MAE, MSE, RMSE, R².
- Cette étape permet de vérifier que le modèle **généralise bien** (et ne fait pas que "mémoriser" les données d'entraînement), et de décider s'il peut être déployé en production ou s'il doit être ajusté (retour éventuel à l'étape 1 ou 2).

## 4. Flowchart du processus de Machine Learning

Le schéma ci-dessous illustre visuellement les grandes étapes du processus, de la collecte des données jusqu'à l'évaluation du modèle (avec la possibilité d'itérer si les résultats ne sont pas satisfaisants).

```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.patches import FancyArrowPatch

steps = [
    "1. Collecte\ndes données",
    "2. Feature\nSelection",
    "3. Model\nSelection",
    "4. Entraînement\ndu modèle",
    "5. Model\nEvaluation",
    "6. Déploiement"
]
colors = ["#95a5a6", "#3498db", "#e67e22", "#9b59b6", "#2ecc71", "#1abc9c"]

fig, ax = plt.subplots(figsize=(14, 4))

box_w, box_h = 1.9, 1.2
gap = 0.5
y = 0

for i, (step, color) in enumerate(zip(steps, colors)):
    x = i * (box_w + gap)
    rect = mpatches.FancyBboxPatch(
        (x, y), box_w, box_h,
        boxstyle="round,pad=0.05,rounding_size=0.08",
        linewidth=2, edgecolor=color, facecolor=color, alpha=0.25
    )
    ax.add_patch(rect)
    ax.text(x + box_w / 2, y + box_h / 2, step, ha='center', va='center',
            fontsize=10, fontweight='bold', color='black')

    if i < len(steps) - 1:
        arrow = FancyArrowPatch(
            (x + box_w, y + box_h / 2), (x + box_w + gap, y + box_h / 2),
            arrowstyle='-|>', mutation_scale=20, linewidth=2, color='black'
        )
        ax.add_patch(arrow)

# Boucle de rétroaction (retour de l'évaluation vers la sélection du modèle)
feedback_x_start = 4 * (box_w + gap) + box_w / 2
feedback_x_end = 2 * (box_w + gap) + box_w / 2
feedback = FancyArrowPatch(
    (feedback_x_start, y - 0.1), (feedback_x_end, y - 0.1),
    connectionstyle="arc3,rad=-0.5",
    arrowstyle='-|>', mutation_scale=18, linewidth=1.5, color='crimson', linestyle='--'
)
ax.add_patch(feedback)
ax.text((feedback_x_start + feedback_x_end) / 2, y - 1.5,
        "Si performance insuffisante : on ajuste\nles features ou on change de modèle",
        ha='center', fontsize=8, color='crimson', style='italic')

ax.set_xlim(-0.5, len(steps) * (box_w + gap))
ax.set_ylim(-2.2, box_h + 0.5)
ax.axis('off')
ax.set_title("Processus de développement d'un modèle de Machine Learning",
              fontsize=13, fontweight='bold', pad=15)

plt.tight_layout()
plt.show()
```

## Conclusion

Le Machine Learning permet aux data analysts de passer d'une analyse **descriptive** (comprendre le passé) à une analyse **prédictive et prescriptive** (anticiper l'avenir et automatiser des décisions). Les trois grands paradigmes — **supervisé**, **non supervisé** et **par renforcement** — répondent chacun à des problématiques différentes selon que les données sont étiquetées ou non, ou qu'il s'agit d'apprendre par interaction avec un environnement.

Quel que soit le type de ML utilisé, le processus de construction d'un modèle reste structuré autour de trois piliers : **choisir les bonnes features**, **choisir le bon modèle**, et **évaluer rigoureusement ses performances** avant tout déploiement — avec un principe clé : itérer tant que les résultats ne sont pas satisfaisants.
